# Attention Model Intuition

**[Модель уваги](https://pradeep-dhote9.medium.com/attention-mechanism-in-deep-learning-1cd0dafd0c21) (Attention Model)** - це потужний механізм в нейронних мережах, який революціонізував сферу NLP. Він дозволяє моделі **фокусуватися на найважливіших частинах вхідних даних**, ігноруючи менш релевантні. Це схоже на те, як людина, читаючи текст, звертає увагу на ключові слова та фрази, ігноруючи менш важливі деталі. Наприклад, при перекладі речення з англійської на українську, ми звертаємо більше уваги на слова, що несуть основний зміст, ніж на прийменники та артиклі.

<img src="https://miro.medium.com/v2/resize:fit:1400/format:webp/1*pEModMOTWOy3bE91ZTfjnQ.png" width="700">



**Переваги моделей уваги:**

* **Покращення точності:** Attention models можуть значно покращити точність моделей NLP, особливо при роботі з довгими текстами або складними залежностями між словами. Це відбувається тому, що модель може зосередитися на найважливіших частинах тексту, не відволікаючись на нерелевантну інформацію.
* **Краще розуміння контексту:** Attention models дозволяють моделі краще зрозуміти контекст, враховуючи взаємозв'язки між словами та фразами. Наприклад, при перекладі речення "Я бачив червону машину", модель уваги може звернути увагу на слово "червону", щоб переконатися, що вона перекладає його як прикметник, а не як іменник.
* **Інтерпретованість:** Attention models дозволяють візуалізувати, на які частини вхідних даних модель звертає увагу, що робить їх більш інтерпретованими. Це може допомогти нам зрозуміти, як модель приймає рішення, та виявити потенційні проблеми з моделлю.
* **Ефективність для довгих послідовностей:** Attention models можуть ефективно обробляти довгі послідовності даних, оскільки вони не потребують обробки всієї послідовності одночасно. Вони можуть фокусуватися на найважливіших частинах послідовності, ігноруючи решту.

# Attention Model: Механізм роботи

Модель уваги обчислює ваги уваги для кожної частини вхідних даних, які визначають, наскільки важлива ця частина для виконання задачі NLP. Ці ваги використовуються для зваженого усереднення вхідних даних, що дозволяє моделі фокусуватися на найважливіших частинах.

**Формула для обчислення ваг уваги:**

$
Attention(Q, K, V) = softmax(\frac{Q \cdot K^T}{\sqrt{d_k}}) \cdot V
$

<img src="https://media.licdn.com/dms/image/v2/D5612AQGsNdHKh10ixg/article-inline_image-shrink_1500_2232/article-inline_image-shrink_1500_2232/0/1710056069036?e=1732752000&v=beta&t=NaQPdiRwuCOTwzFZ9738FbGmZOZDOEuXgdN3aAEOinE" width="700">
**Пояснення:**

* **Q (Query):** Вектор запиту, який представляє те, що модель шукає у вхідних даних. Наприклад, при перекладі речення, вектор запиту може представляти слово, яке ми хочемо перекласти.
* **K (Key):** Вектори ключів, які представляють різні частини вхідних даних. Наприклад, при перекладі речення, вектори ключів можуть представляти всі слова в реченні.
* **V (Value):** Вектори значень, які містять інформацію про кожну частину вхідних даних. Наприклад, при перекладі речення, вектори значень можуть представляти вектори word embeddings для кожного слова в реченні.
* **d_k:** Розмірність векторів ключів.
* **softmax:** Функція softmax нормалізує ваги уваги, щоб їх сума дорівнювала 1. Це гарантує, що ваги уваги представляють ймовірність того, що кожна частина вхідних даних є релевантною для запиту.

<img src="https://media.beehiiv.com/cdn-cgi/image/fit=scale-down,format=auto,onerror=redirect,quality=80/uploads/asset/file/540119f8-93a5-48d6-b44d-ecfc301e2330/sa.jpeg?t=1700819627" width="700">

**Процес обчислення уваги:**

1. **Обчислення подібності:** Для кожного вектора ключа `K` обчислюється його подібність до вектора запиту `Q`. Це робиться шляхом обчислення скалярного добутку `Q` та `K`.
2. **Нормалізація:** Подібності нормалізуються за допомогою функції softmax, щоб отримати ваги уваги.
3. **Зважене усереднення:** Вектори значень `V` зважуються за допомогою ваг уваги та усереднюються.

**Приклад:**

Уявіть собі, що ми хочемо перекласти речення "The cat sat on the mat" з англійської на українську.

1. **Вектор запиту `Q`:** Може представляти слово "cat".
2. **Вектори ключів `K`:** Можуть представляти всі слова в реченні: "The", "cat", "sat", "on", "the", "mat".
3. **Вектори значень `V`:** Можуть представляти вектори word embeddings для кожного слова в реченні.

Модель уваги обчислить подібність вектора запиту "cat" до кожного вектора ключа. Найбільша подібність буде у вектора ключа "cat", тому вага уваги для цього слова буде найбільшою. Модель потім зважить вектори значень за допомогою ваг уваги та усереднить їх. Це дозволить моделі фокусуватися на слові "cat" та його контексті при перекладі речення.

## Transformers architecture

Архітектура **Transformer**, представлена у 2017 році в статті "Attention Is All You Need", стала революційним проривом в області обробки природної мови (NLP), а також знайшла застосування в інших сферах, таких як комп'ютерний зір та обробка аудіо. Вона відмовилася від традиційних рекурентних нейронних мереж (RNN) та згорткових нейронних мереж (CNN) на користь механізму уваги, що дозволило досягти значно кращих результатів та ефективності.

**Основні переваги архітектури Transformer:**

* **Паралелізація:** Transformer може обробляти всю вхідну послідовність одночасно, що робить його значно швидшим за RNN, які обробляють послідовність покроково.
* **Довгострокові залежності:** Transformer може ефективно моделювати довгострокові залежності між словами в послідовності, завдяки механізму уваги, який може "бачити" всю послідовність одночасно.
* **Універсальність:** Transformer може бути використаний для широкого спектру завдань NLP, таких як машинний переклад, генерація тексту, аналіз настроїв та багато іншого.



**Структура Transformer:**

Архітектура Transformer складається з двох основних частин:

* **Енкодер:** Енкодер відповідає за обробку вхідної послідовності та створення представлення, яке кодує всю інформацію з послідовності.
* **Декодер:** Декодер використовує представлення, створене енкодером, для генерації вихідної послідовності.


<img src="https://datasciencedojo.com/wp-content/uploads/Architecture-of-transformer-models-1030x579.jpg" width="1000">



**Енкодер:**

Енкодер складається з кількох однакових шарів, кожен з яких містить два підшари:

* **Multi-Head Self-Attention:** Цей підшар дозволяє моделі звертати увагу на різні частини вхідної послідовності та визначати, які слова є найважливішими для кожного слова.

<img src="https://substackcdn.com/image/fetch/f_auto,q_auto:good,fl_progressive:steep/https%3A%2F%2Fsubstack-post-media.s3.amazonaws.com%2Fpublic%2Fimages%2Ff406d55e-990a-4e3b-be82-d966eb74a3e7_1766x1154.png" width="700">

* **Position-wise Feed-Forward Network:** Цей підшар обробляє вихід механізму само-уваги, застосовуючи до нього нелінійне перетворення.

<img src="https://miro.medium.com/v2/resize:fit:1906/1*1l5JbeGfEGh2oxjI8koHdQ.png" width="700">


**Декодер:**

Декодер також складається з кількох однакових шарів, кожен з яких містить три підшари:

* **Багатоголовий механізм маскированої само-уваги (Multi-Head Masked Self-Attention):** Цей підшар схожий на механізм само-уваги в енкодері, але він "маскує" майбутні слова в вихідній послідовності, щоб модель не "підглядала" на них під час генерації.
* **Багатоголовий механізм уваги енкодер-декодер (Multi-Head Encoder-Decoder Attention):** Цей підшар дозволяє моделі звертати увагу на представлення, створене енкодером, щоб отримати інформацію з вхідної послідовності.
* **Позиційно-залежний Feed-Forward шар (Position-wise Feed-Forward Network):** Цей підшар обробляє вихід механізму уваги енкодер-декодер, застосовуючи до нього нелінійне перетворення.




**Позиційне кодування (Positional Encoding):**

Оскільки Transformer не використовує рекурентність, він потребує механізму для врахування порядку слів у послідовності. Для цього використовується позиційне кодування, яке додає до векторів word embeddings інформацію про позицію слова в послідовності.

# Transformers library

<img src="https://miro.medium.com/v2/resize:fit:1400/1*PMlT7m9nMigseMmJcB4ddw.png" width="700">

Бібліотека Transformers, розроблена Hugging Face, є потужним інструментом для роботи з моделями Transformer, які є основою багатьох сучасних моделей NLP.

**Переваги бібліотеки Transformers:**

* **Широкий вибір попередньо навчених моделей:** Бібліотека надає доступ до сотень попередньо навчених моделей для різних завдань NLP, таких як BERT, GPT, RoBERTa, XLNet, T5 та багато інших. Ці моделі були навчені на величезних наборах даних і можуть бути використані для різних завдань без необхідності навчання з нуля.
* **Простота використання:** Бібліотека має простий та інтуїтивно зрозумілий API, що дозволяє легко завантажувати, налаштовувати та використовувати попередньо навчені моделі. Це робить бібліотеку доступною для широкого кола користувачів, включаючи тих, хто не має глибоких знань про глибоке навчання.
* **Підтримка різних фреймворків:** Бібліотека підтримує TensorFlow, PyTorch та JAX, що дозволяє використовувати моделі з будь-яким з цих фреймворків. Це дає користувачам гнучкість у виборі фреймворку, який їм найкраще підходить.
* **Активна спільнота:** Бібліотека має велику та активну спільноту розробників та дослідників, що постійно розширюють її можливості та додають нові моделі. Це гарантує, що бібліотека буде актуальною та корисною для тривалого часу.
* **Інструменти для тонкого налаштування та оцінювання:** Бібліотека надає інструменти для тонкого налаштування попередньо навчених моделей на нових наборах даних, а також для оцінювання продуктивності моделей. Це спрощує процес адаптації моделей до конкретних завдань.

## Основні можливості бібліотеки Transformers:



### **Завантаження попередньо навчених моделей:**

```python
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained("bert-base-uncased")
```

**Пояснення:**

* `AutoModelForSequenceClassification` - це клас, який автоматично завантажує модель для класифікації послідовностей (наприклад, класифікація тексту) з заданою архітектурою (в даному випадку BERT).
* `from_pretrained("bert-base-uncased")` - завантажує попередньо навчену модель BERT з Hugging Face Model Hub. "bert-base-uncased" - це назва моделі.



### **Тонке налаштування моделей:**

```python
from transformers import TrainingArguments, Trainer

training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=3,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=64,
    warmup_steps=500,
    weight_decay=0.01,
    logging_dir="./logs",
    logging_steps=10,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
)

trainer.train()
```

**Пояснення:**

* `TrainingArguments` - клас, який визначає аргументи для навчання моделі, такі як кількість епох, розмір пакету, швидкість навчання та ін.
* `Trainer` - клас, який спрощує процес навчання моделі. Він автоматично виконує оптимізацію, оцінювання та збереження моделі.



### **Використання моделей для задач NLP:**

```python
from transformers import pipeline

classifier = pipeline("text-classification", model=model, tokenizer=tokenizer)
result = classifier("Це текст про науку.")
print(result)
```

**Пояснення:**

* `pipeline` - це функція, яка створює об'єкт pipeline для виконання задачі NLP.
* "text-classification" - вказує на те, що ми хочемо використовувати pipeline для класифікації тексту.
* `model` та `tokenizer` - це об'єкти моделі та токенізатора, які ми хочемо використовувати.



### **Токенізація та обробка тексту:**

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
inputs = tokenizer("Hello, world!", return_tensors="pt")
```

**Пояснення:**

* `AutoTokenizer` - це клас, який автоматично завантажує токенізатор для заданої архітектури (в даному випадку BERT).
* `tokenizer("Hello, world!", return_tensors="pt")` - токенізує текст "Hello, world!" та повертає тензори PyTorch.

### Torch: Фреймворк для глибокого навчання

PyTorch - це потужний фреймворк для глибокого навчання з відкритим кодом, розроблений Facebook AI Research. PyTorch широко використовується в NLP для створення та навчання моделей глибокого навчання.

**Переваги PyTorch:**

* **Динамічні обчислювальні графи:** PyTorch використовує динамічні обчислювальні графи, що дозволяє гнучко змінювати структуру моделі під час виконання. Це означає, що ви можете створювати моделі з розгалуженнями, циклами та іншими складними структурами, які неможливо створити за допомогою статичних графів, як у TensorFlow.
* **Простота використання:** PyTorch має простий та інтуїтивно зрозумілий API, що робить його легким у вивченні та використанні. Це особливо важливо для новачків у глибокому навчанні.
* **Широка підтримка:** PyTorch підтримується великою спільнотою розробників та має багато доступних ресурсів та документації. Це означає, що ви можете знайти багато прикладів коду, навчальних посібників та форумів, де можна отримати допомогу.
* **Ефективність:** PyTorch є досить ефективним фреймворком, який може працювати як на CPU, так і на GPU. Це дозволяє навчати моделі на великих наборах даних за розумний час.
* **Налагодження:** PyTorch надає потужні інструменти для налагодження моделей, що спрощує процес виявлення та виправлення помилок.




**Основні можливості PyTorch:**

* **Тензори:** PyTorch використовує тензори для представлення даних, які схожі на NumPy arrays, але можуть бути оброблені на GPU. Тензори є основним будівельним блоком для моделей глибокого навчання.
* **Автоматичне диференціювання:** PyTorch може автоматично обчислювати градієнти, що є важливим для навчання моделей глибокого навчання. Градієнти використовуються для оновлення ваг моделі під час навчання.
* **Модулі:** PyTorch надає модулі для створення складних нейронних мереж, таких як лінійні шари, згорткові шари та рекурентні шари. Модулі є готовими блоками, які можна використовувати для створення моделей.
* **Оптимізатори:** PyTorch надає оптимізатори для налаштування ваг моделі під час навчання, такі як SGD, Adam та RMSprop. Оптимізатори визначають, як оновлювати ваги моделі на основі градієнтів.
* **Функції втрат:** PyTorch надає функції втрат для оцінки продуктивності моделі під час навчання. Функції втрат визначають, наскільки добре модель передбачає правильні відповіді.
* **Набори даних:** PyTorch надає інструменти для створення та завантаження наборів даних. Набори даних використовуються для навчання та оцінювання моделей.


````python
import torch
import torch.nn as nn

class SimpleNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleNN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

# Створення моделі
model = SimpleNN(input_size=10, hidden_size=20, output_size=5)

# Вхідні дані
input_data = torch.randn(1, 10)

# Прогін моделі
output = model(input_data)

print(output)
````

**Пояснення:**

* `SimpleNN` - це клас, який визначає нашу просту нейронну мережу.
* `__init__` - це конструктор класу, який ініціалізує шари мережі.
* `forward` - це метод, який визначає, як дані проходять через мережу.
* `nn.Linear` - це лінійний шар, який виконує лінійне перетворення даних.
* `nn.ReLU` - це функція активації ReLU, яка додає нелінійність до мережі.
* `torch.randn` - створює тензор з випадковими числами з нормального розподілу.

## Transfer Learning з BERT

Transfer learning з BERT дозволяє адаптувати попередньо навчену модель BERT до конкретного завдання NLP, використовуючи невеликий набір даних. Це дозволяє досягти високої точності, навіть з обмеженими даними.

<img src="https://jalammar.github.io/images/bert-transfer-learning.png" width="700">

````python
!pip install datasets
````

````python
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer, AutoTokenizer

# Завантаження попередньо навченої моделі та токенізатора
model = AutoModelForSequenceClassification.from_pretrained("distilbert/distilbert-base-uncased", num_labels=2)
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")

# Завантаження набору даних з Hugging Face
from datasets import load_dataset

dataset = load_dataset("imdb")

# Підготовка даних
def preprocess_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True)

encoded_dataset = dataset.map(preprocess_function, batched=True)
````

````python
# Визначення аргументів навчання
training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=1,
    per_device_train_batch_size=24,
    per_device_eval_batch_size=32,
    warmup_steps=500,
    weight_decay=0.01,
    logging_dir="./logs",
    logging_steps=10,
)

# Створення тренера
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=encoded_dataset["train"],
    eval_dataset=encoded_dataset["test"],
)
````

````python
# Навчання моделі
trainer.train()
````

````python
import matplotlib.pyplot as plt

# Assuming you have a list of training loss values and evaluation loss values
# obtained during training
train_losses = trainer.state.log_history  # Or however you store your training loss

# Extract the loss values from the log history
train_losses = [entry['loss'] for entry in train_losses if 'loss' in entry]

# Extract evaluation loss values (if available)
eval_losses = []
for entry in trainer.state.log_history:
    if 'eval_loss' in entry:
        eval_losses.append(entry['eval_loss'])

# Plot the learning curves
plt.plot(train_losses, label='Training Loss')
if eval_losses:
    plt.plot(eval_losses, label='Evaluation Loss')

plt.xlabel('Epochs/Steps')
plt.ylabel('Loss')
plt.title('Learning Curves')
plt.legend()
plt.show()
````

````python
import numpy as np
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

def compute_metrics(pred):
  labels = pred.label_ids
  preds = pred.predictions.argmax(-1)
  return {"accuracy": (preds == labels).mean()}


# Get predictions for the test dataset
predictions = trainer.predict(encoded_dataset["test"])
# Extract the predicted labels and true labels
predicted_labels = np.argmax(predictions.predictions, axis=1)
true_labels = predictions.label_ids

# Compute the confusion matrix
cm = confusion_matrix(true_labels, predicted_labels)

# Display the confusion matrix using seaborn
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.title("Confusion Matrix")
plt.xlabel("Predicted Labels")
plt.ylabel("True Labels")
plt.show()
````

```python
import streamlit as st
from openai import OpenAI


st.title("ChatGPT-based application")

# Set OpenAI API key from Streamlit secrets
client = OpenAI(api_key=st.secrets["OPENAI_API_KEY"])

# Initialize chat history
if "messages" not in st.session_state:
    st.session_state.messages = []

# Display chat messages from history on app rerun
for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# Accept user input
if prompt := st.chat_input("Type here ..."):
    
    # Add user message to chat history
    st.session_state.messages.append({"role": "user", "content": prompt})
    
    # Display user message in chat message container
    with st.chat_message("user"):
        st.markdown(prompt)

    # Display assistant response in chat message container
    with st.chat_message("assistant"):
        stream = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=[
                {"role": m["role"], "content": m["content"]}
                for m in st.session_state.messages
            ],
            stream=True,
        )
        response = st.write_stream(stream)
    
    st.session_state.messages.append({"role": "assistant", "content": response})

```

## Створення API для класифікації рецензій на фільми з FastAPI

**Мета:** Створити API для класифікації рецензій на фільми з використанням FastAPI, бібліотеки Transformers та попередньо навченої моделі DistilBERT.

**Кроки:**

1. **Встановлення бібліотек:**

```bash
pip install fastapi uvicorn transformers torch
```

2. **Створення файлу `main.py`:**

```python
import uvicorn
from pydantic import BaseModel
from fastapi import FastAPI, HTTPException
from transformers import (
    AutoModelForSequenceClassification,
    AutoTokenizer,
    TextClassificationPipeline
)


# Load fine-tuned model and tokenizer
path = './imdb_dbert'
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")
model = AutoModelForSequenceClassification.from_pretrained(path)
pipe = TextClassificationPipeline(model=model, tokenizer=tokenizer)

# FastAPI application creation
app = FastAPI(
    title="CineSent API",
    description="API for movie review classification",
    version="0.1.0",
)

# Request model
class ReviewRequest(BaseModel):
    text: str

# Classification endpoint
@app.post("/classify_review/", response_model=str)
async def classify_review(review: ReviewRequest):
    """
    Classify review either as "POSITIVE" or "NEGATIVE".
    """
    try:
        result = pipe(review.text)[0]
        print(result)

        match result["label"]:
            case "LABEL_0":
                response = "NEGATIVE"
            case "LABEL_1":
                response = "POSITIVE"
            case _:
                response = "POSITIVE"

        return response
    except Exception as e:
        raise HTTPException(status_code=500, detail=f"Error during classification: {e}")


# Launch part
if __name__ == "__main__":
    uvicorn.run(app, host="0.0.0.0", port=8000)
```

**Пояснення:**

* **Завантаження моделі:** Попередньо навчена модель DistilBERT та токенізатор завантажуються за допомогою бібліотеки Transformers.
* **FastAPI додаток:** Створюється екземпляр FastAPI додатку.
* **Модель даних:** Визначається модель даних `ReviewInput` з використанням Pydantic, що вимагає єдине поле `text` для рецензії.
* **Ендпоінт `/classify_review/`:**
    * Декорується `@app.post("/classify_review/", response_model=str)`, що означає, що він обробляє POST-запити на цей URL і повертає рядок.
    * Приймає об'єкт `ReviewInput` як вхідні дані.
    * Використовує завантажену модель для класифікації рецензії.
    * Повертає мітку ("POSITIVE" або "NEGATIVE") як результат.
    * Обробляє можливі винятки під час класифікації, повертаючи HTTPException з кодом 500.
* **Запуск Uvicorn:** Додаток запускається за допомогою Uvicorn ASGI сервера на порту 8000.

**Запуск API:**

1. Збережіть код у файлі `main.py`.
2. Відкрийте термінал та перейдіть до директорії з файлом `main.py`.
3. Запустіть команду `uvicorn main:app --reload`, щоб запустити сервер з автоматичним перезавантаженням при зміні коду.

**Використання API:**

Після запуску API, ви можете відправляти POST-запити на URL `http://127.0.0.1:8000/classify_review/` з JSON тілом, що містить поле `text` з рецензією на фільм.

**Приклад запиту:**

```json
{
  "text": "This movie was absolutely fantastic! The acting was phenomenal and the story kept me on the edge of my seat the entire time."
}
```

**Відповідь:**

```json
"POSITIVE"
```

**Висновок:**

Створено API для класифікації рецензій на фільми, який може бути використаний для автоматизації процесу аналізу настроїв та надання цінної інформації для "CineSent".


## Обертання API в Docker

Для зручності розгортання та ізоляції залежностей, ми обернемо наш API для класифікації рецензій на фільми в простий Docker контейнер.

### Docker: Контейнери для ваших додатків

Уявіть собі, що ви розробляєте додаток. Вам потрібно встановити певні бібліотеки, налаштувати середовище, можливо, навіть встановити окрему базу даних. А потім вам потрібно перенести цей додаток на інший комп'ютер, наприклад, на сервер для розгортання. Чи гарантовано, що все працюватиме так само, як на вашій машині?  

**Тут на допомогу приходить Docker.**

**Що таке Docker?**

Docker - це платформа з відкритим кодом, яка дозволяє упаковувати додатки разом з усіма їх залежностями (бібліотеками, налаштуваннями, іншими компонентами) у **контейнери**.

<img src="https://d2908q01vomqb2.cloudfront.net/ca3512f4dfa95a03169c5a670a4c91a19b3077b4/2020/03/11/prasanna_lead-image_2020_03-11_750.png" width="1000">

**Контейнер Docker** - це легкий, портативний та самодостатній пакет програмного забезпечення, який містить все необхідне для запуску додатку.

**Як працює Docker?**

Docker використовує **образи (images)** та **контейнери (containers)**.

* **Образ Docker:** Це шаблон, який містить інструкції для створення контейнера. Він визначає операційну систему, бібліотеки, налаштування та інші компоненти, необхідні для запуску додатку.
* **Контейнер Docker:** Це екземпляр образу, який запущений і працює. Ви можете створювати, запускати, зупиняти та видаляти контейнери на основі одного образу.

<img src="https://res.cloudinary.com/practicaldev/image/fetch/s--kcatXkgu--/c_limit%2Cf_auto%2Cfl_progressive%2Cq_auto%2Cw_800/https://dev-to-uploads.s3.amazonaws.com/uploads/articles/9jxn6fjons4ggrfrnxek.png" width="700">



**Переваги Docker:**

* **Ізоляція:** Контейнери Docker ізольовані від системи хоста та один від одного, що забезпечує узгодженість середовища розробки та розгортання. Це означає, що додаток працюватиме однаково незалежно від того, де він запущений.
* **Портативність:** Docker контейнери можуть бути легко перенесені на інші машини, що спрощує розгортання. Ви можете створити контейнер на своєму комп'ютері, а потім запустити його на сервері або в хмарі без будь-яких змін.
* **Масштабованість:** Docker спрощує масштабування додатку, дозволяючи запускати кілька контейнерів на різних серверах.
* **Ефективність:** Контейнери Docker є легкими та швидкими, що робить їх ідеальними для розгортання мікросервісів.
* **Контроль версій:** Docker образи можна зберігати в репозиторіях, що дозволяє відстежувати зміни та контролювати версії додатку.

**Використання Docker:**

Docker використовується розробниками, системними адміністраторами та DevOps інженерами для:

* **Розробки:** Створення ізольованих середовищ розробки для різних проектів.
* **Тестування:** Запуск тестів у контрольованому середовищі.
* **Розгортання:** Розгортання додатків на серверах, в хмарі та інших середовищах.




**Кроки:**

1. **Створіть файл `Dockerfile` в директорії проекту:**

```dockerfile
FROM python:3.12

WORKDIR /app

COPY requirements.txt .
RUN pip install --no-cache-dir --upgrade -r requirements.txt

COPY . .

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

**Пояснення `Dockerfile`:**

* **`FROM python:3.12`:** Використовуємо базовий образ Docker з Python 3.12.
* **`WORKDIR /app`:** Встановлюємо робочу директорію всередині контейнера.
* **`COPY requirements.txt .`:** Копіюємо файл `requirements.txt` в робочу директорію.
* **`RUN pip install ...`:** Встановлюємо залежності з файлу `requirements.txt`.
* **`COPY . .`:** Копіюємо весь код проекту в робочу директорію.
* **`CMD [...]`:** Визначаємо команду, яка буде запущена при запуску контейнера - запуск Uvicorn ASGI сервера.

2. **Створіть файл `requirements.txt` в директорії проекту:**

```
fastapi
uvicorn
transformers
pydantic
torch
```

3. **Побудуйте Docker образ:**

```bash
docker build -t cinesent-api .
```

**Пояснення:**

* `-t cinesent-api`: Встановлюємо назву образу `cinesent-api`.
* `.`: Вказуємо на поточну директорію, де знаходиться `Dockerfile`.

4. **Запустіть Docker контейнер:**

```bash
docker run -d -p 8000:8000 cinesent-api
```

**Пояснення:**

* `-d`: Запускаємо контейнер у фоновому режимі.
* `-p 8000:8000`: Прокидуємо порт 8000 з контейнера на порт 8000 хоста.
* `cinesent-api`: Назва образу, який ми побудували.

**Висновок:**

Тепер ваш API для класифікації рецензій на фільми запущено в Docker контейнері. Ви можете отримати доступ до API за URL `http://127.0.0.1:8000/classify_review/`, як і раніше.

**Переваги використання Docker:**

* **Ізоляція:** Docker ізолює залежності додатку від системи хоста, що забезпечує узгодженість середовища розробки та розгортання.
* **Портативність:** Docker контейнери можуть бути легко перенесені на інші машини, що спрощує розгортання.
* **Масштабованість:** Docker спрощує масштабування додатку, дозволяючи запускати кілька контейнерів на різних серверах.
